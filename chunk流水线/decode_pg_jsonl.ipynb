# JSONL 的 pg 字段解码
修改下方两个路径，然后运行单元格。只解码 `pg` 额外的 JSON 字符串编码一层；普通 HTML 和其他字段保持不变。

输出必须使用新文件名，避免覆盖原文件；中途报错可能留下部分输出，修正后请换新输出文件名。输出仍是 JSONL，文本中的转义应在读取 JSON 后检查。仅依赖 Python 标准库。

```python
import json

# 修改为你的输入文件路径和新输出文件路径
input_path = "output_pg_2.jsonl"
output_path = "output_pg_2_decoded.jsonl"

total = decoded_count = 0

with open(input_path, encoding="utf-8-sig") as fin, \
     open(output_path, "x", encoding="utf-8") as fout:
    for line in fin:
        if not line.strip():
            continue

        record = json.loads(line)
        pg = record.get("pg")

        if isinstance(pg, str):
            try:
                decoded = json.loads(pg)
                if isinstance(decoded, str):
                    record["pg"] = decoded
                    decoded_count += 1
            except json.JSONDecodeError:
                pass  # 已经是普通 HTML，保持原样

        fout.write(json.dumps(record, ensure_ascii=False) + "\n")
        total += 1

print(f"总计 {total} 条，额外解码 {decoded_count} 条")
print(f"输出：{output_path}")
```
